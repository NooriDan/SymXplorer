```python
from symxplorer.designer_tools.visualizer import ExplorationPoint, DesignSpaceExploration, AnalysisSuite

# Example exploration data
points = [
    ExplorationPoint({"w": 1.0, "l": 0.1}, {"score": 0.8, "ugf": 120e6, "pm": 65}),
    ExplorationPoint({"w": 2.0, "l": 0.2}, {"score": 0.9, "ugf": 140e6, "pm": 72}),
    ExplorationPoint({"w": 3.0, "l": 0.3}, {"score": 0.7, "ugf": 110e6, "pm": 58}),
    ExplorationPoint({"w": 4.0, "l": 0.4}, {"score": 0.85, "ugf": 130e6, "pm": 68}),
]

exploration = DesignSpaceExploration(points)
analysis = AnalysisSuite(exploration)

# Interactive plots
analysis.plot_sensitivity("score")
analysis.plot_slice("x_w", "score")
analysis.plot_contour("x_w", "x_l", "score")
analysis.plot_progression("score")
analysis.plot_parallel(["score", "ugf", "pm"])
```
